## Tool Choice 제어 및 Function Calling 종합 실습: 날씨 → 활동 추천 자동 체인

**실습 목표**
- OpenAI Function Calling의 실제 사용 흐름을 이해한다.
- 모델이 **여러 개의 툴을 순차적으로 호출하는 과정(Chained Tool Calling)** 을 학습한다.
- 기온 조회 → 활동 추천 → 최종 자연어 응답까지 이어지는 전체 파이프라인을 구현한다.

**실습 구성**

1. **get_weather**
   - 입력: `city`
   - 출력: 현재 기온
   - 실제 API(Open-Meteo)로 기온을 조회
   - Step 1에서는 이 툴만 사용 가능하도록 `tool_choice` 제한

2. **recommend_activity**
   - 입력: `temperature`
   - 출력: 기온 구간별 추천 활동
   - Step 2에서만 호출되도록 `tool_choice`로 제어

3. **Function Calling 체인 (총 3회 호출)**
   - **Step 1 — get_weather 호출**
     - 입력: “서울 날씨 + 활동 추천”
     - `tool_choice`로 `get_weather`만 허용 → 모델이 자동으로 tool_call 생성
     - instructor가 툴 실행 후 결과를 모델에 재전달
   - **Step 2 — recommend_activity 호출**
     - `tool_choice`로 `recommend_activity`만 허용 → 모델이 tool_call 생성
     - instructor가 결과를 다시 모델에 전달
   - **Step 3 — 최종 답변 생성**
     - `tool_choice = "none"`
     - 모델이 도구 없이 자연어로 마무리 답변
     - 예: “현재 기온은 X도이며, Y 같은 활동을 추천합니다.”

```python
import json
import requests
from openai import OpenAI

client = OpenAI()

# =====================================================================
# 1. Tool 정의
# =====================================================================

# get_weather 툴
weather_tool = {
    "type": "function",
    "name": "get_weather",
    "description": "지원 가능한 도시의 현재 기온을 조회합니다.",
    "parameters": {
        "type": "object",
        "properties": {
            "city": {
                "type": "string",
                "description": "조회할 도시 이름입니다. (예: '서울', '뉴욕')"
            }
        },
        "required": ["city"],
    },
}

# recommend_activity 툴
recommend_activity_tool = {
    "type": "function",
    "name": "recommend_activity",
    "description": "기온을 기반으로 적절한 활동을 추천합니다.",
    "parameters": {
        "type": "object",
        "properties": {
            "temperature": {
                "type": "number",
                "description": "섭씨 기온"
            }
        },
        "required": ["temperature"],
    },
}

#dummy tool
web_search_tool = {
    "type": "web_search"
    }

all_tools = [weather_tool, recommend_activity_tool, web_search_tool]


# =====================================================================
# 2. Tool 함수 구현
# =====================================================================

CITY_COORDS = {
    "서울":  (37.5665, 126.9780),
    "부산":  (35.1796, 129.0756),
    "뉴욕":  (40.7128, -74.0060),
    "도쿄":  (35.6895, 139.6917),
    "파리":  (48.8566, 2.3522),
    "두바이": (25.2048, 55.2708),
}

def get_weather(city: str):
    if city not in CITY_COORDS:
        return {"error": f"'{city}'는 현재 지원하지 않는 도시입니다."}

    lat, lon = CITY_COORDS[city]

    url = "https://api.open-meteo.com/v1/forecast"
    data = requests.get(url, params={
        "latitude": lat,
        "longitude": lon,
        "current_weather": True,
    }).json()

    temp = data["current_weather"]["temperature"]
    return {"city": city, "temperature": temp}

def recommend_activity(temperature):
    if temperature >= 25:
        return {"activity": "시원한 음료 마시면서 공원 산책하기"}
    elif temperature >= 15:
        return {"activity": "한강에서 자전거 타기"}
    elif temperature >= 5:
        return {"activity": "따뜻한 카페에서 책 읽기"}
    else:
        return {"activity": "집에서 넷플릭스 보기"}


# =====================================================================
# 3. 사용자 입력
# =====================================================================

input_list = [
    {"role": "user", "content": "지금 서울 날씨 알려줘. 그리고 그 날씨에서 하면 좋을 활동도 추천해줘."}
]


# =====================================================================
# 4. First Call — get_weather 호출
# =====================================================================

response1 = client.responses.create(
    model="gpt-6-luna",
    tools=all_tools,
    input=input_list,
    tool_choice={
        "type": "allowed_tools",
        "mode": "required",
        "tools": [{"type": "function", "name": "get_weather"}],
    }
)

input_list += response1.output

# Tool Call 처리
for item in response1.output:
    if item.type == "function_call" and item.name == "get_weather":
        args = json.loads(item.arguments)
        result = get_weather(args["city"])
        print("🛠️ get_weather:", args, "→", result)

        input_list.append({
            "type": "function_call_output",
            "call_id": item.call_id,
            "output": json.dumps(result, ensure_ascii=False),
        })


# =====================================================================
# 5. Second Call — recommend_activity 호출
# =====================================================================

response2 = client.responses.create(
    model="gpt-6-luna",
    tools=all_tools,
    input=input_list,
    tool_choice={
        "type": "allowed_tools",
        "mode": "required",
        "tools": [{"type": "function", "name": "recommend_activity"}],
    }
)

input_list += response2.output

for item in response2.output:
    if item.type == "function_call" and item.name == "recommend_activity":
        args = json.loads(item.arguments)
        result = recommend_activity(args["temperature"])
        print("🛠️ recommend_activity:", args, "→", result)

        input_list.append({
            "type": "function_call_output",
            "call_id": item.call_id,
            "output": json.dumps(result, ensure_ascii=False),
        })


# =====================================================================
# 6. Third Call — 최종 자연어 응답 생성 (더 이상 도구 사용 금지)
# =====================================================================

final = client.responses.create(
    model="gpt-6-luna",
    tools=all_tools,
    input=input_list,
    instructions="지금까지 도구에서 얻은 정보만 사용해서 자연어로 설명하세요.",
    tool_choice="none",
)

print("\n🌤 최종 답변:")
print(final.output_text)
```
